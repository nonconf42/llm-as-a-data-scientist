```python
def hyp_res_to_text(hypothesis_results):
    res = ''
    for tup in hypothesis_results:
        tmp = tup[0]+': ' + str(bool(tup[1]))
        res += (tmp + '\n')
    return res
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split

def split_and_save_dataset(dataset_name):
    """
    Splits the dataset into training and testing sets and saves them as CSV files.
    
    Args:
        dataset_name (str): The name of the dataset file (CSV format).
    """
    try:
        # Load the dataset
        df = pd.read_csv(dataset_name)
        print(f"Dataset loaded successfully with shape: {df.shape}")

        # Split the dataset into 80% train and 20% test
        train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)
        print(f"Train set shape: {train_df.shape}, Test set shape: {test_df.shape}")

        # Save the dataframes as CSV files
        train_df.to_csv("train.csv", index=False)
        test_df.to_csv("test.csv", index=False)
        print("Dataframes saved successfully as 'train.csv' and 'test.csv'.")

    except FileNotFoundError:
        print(f"Error: The file '{dataset_name}' was not found.")
    except Exception as e:
        print(f"An error occurred: {e}")


dataset_name = 'Churn.csv'
split_and_save_dataset(dataset_name)
```

```python
from matplotlib import pyplot as plt
import matplotlib.patches as patches

# Define the text content for each box
content = [
    ("Hypothesis Prompt", [
        "1. Churn Dataset Description",
        "2. Columns description and column statistics",
        "3. Instruction prompt to generate hypotheses"
    ]),
    ("Hypotheses generated by gpt-4o (Churn Dataset)", [
        "1. Active bank members show a retention rate 20% higher than inactive members, "
        "indicating that improving engagement levels for inactive members could improve overall retention.",
        "2. The observed 18% higher exit rate for women suggests banks need to focus on understanding female customer needs better.",
        "3. Exit rates drop to 11% for customers with more than two products compared to 18% for those with fewer, "
        "emphasizing the value of cross-project strategies like bundling or offering complementary services to boost product engagement."
    ]),
    ("COTs generated by gpt-4o", [
        "COT 1:",
        "1. Define membership activity by separating customers into 'active' and 'inactive' groups (IsActiveMember = 1 and 0, respectively).",
        "2. Calculate the retention rate for each group measured by the proportion of customers who have not exited.",
        "3. Use a chi-squared test to see if active members have a significantly higher retention rate than inactive ones.",
        "4. Code the function to carry out these calculations and tests.",
        "5. Analyze and interpret the results to suggest improvements in customer engagement strategies."
    ]),
    ("Code generated by GPT-4", [
        "def test(dataset):",
        "    import pandas as pd",
        "    from scipy.stats import chi2_contingency",
        "",
        "    more_than_two = dataset[dataset['NumOfProducts'] > 2]['Exited'].mean()",
        "    two_or_fewer = dataset[dataset['NumOfProducts'] <= 2]['Exited'].mean()",
        "",
        "    obs = pd.DataFrame({'Exited': [more_than_two, two_or_fewer], ",
        "                        'Retained': [1 - more_than_two, 1 - two_or_fewer]})",
        "    ",
        "    chi2, p, _, _ = chi2_contingency(obs)",
        "    ",
        "    result = 1 if p < 0.05 else 0",
        "    return result"
    ]),
    ("Hypothesis Testing Result", [
        "1 - if hypothesis accepted",
        "0 - otherwise"
    ])
]

# Define box properties
box_width = 10
box_height = 1.5
gap = 0  # No gap between boxes
font_size = 10

# Create the plot
fig, ax = plt.subplots(figsize=(10, len(content) * (box_height + gap)))

y = len(content) * (box_height + gap)  # Starting y position

for title, lines in content:
    # Draw the box
    rect = patches.Rectangle((0, y - box_height), box_width, box_height, linewidth=1,
                              edgecolor='blue', facecolor='lightblue')
    ax.add_patch(rect)
    # Add title text
    ax.text(0.5, y - 0.5 * box_height, title, fontsize=font_size, fontweight='bold',
            verticalalignment='center', horizontalalignment='center')
    # Add content text
    for i, line in enumerate(lines):
        ax.text(0.2, y - box_height - 0.25 - i * 0.3, line, fontsize=font_size, verticalalignment='top')
    y -= (box_height + gap + len(lines) * 0.3)  # Adjust y position for the next box

# Set limits and remove axes
ax.set_xlim(-1, box_width + 1)
ax.set_ylim(0, len(content) * (box_height + gap))
ax.axis('off')

# Save as PNG
plt.show()

```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the data
data = pd.read_csv('new_reg_results.csv')

# Extract relevant columns
data = data[['Dataset', 'LLM', 'Performance of M1', 'Performance of M2', 'Performance of M3']]

# Function to extract RMSE from performance string
def extract_rmse(performance):
    return eval(performance)[1]

# Apply the function to extract RMSE
for col in ['Performance of M1', 'Performance of M2', 'Performance of M3']:
    data[col] = data[col].apply(extract_rmse)

# Find the minimum RMSE for each LLM and Dataset combination
best_rmse = data.groupby(['Dataset', 'LLM']).min().reset_index()

# Pivot the data to have LLM as rows and Datasets as columns
pivot_data = best_rmse.pivot(index='LLM', columns='Dataset', values=['Performance of M1', 'Performance of M2', 'Performance of M3'])

# Aggregate to find the best RMSE across all regressors (M1, M2, M3)
pivot_data = pivot_data.min(axis=1)

# Plot the data
pivot_data.plot(kind='bar', figsize=(12, 6))
plt.title('Best RMSE for Each LLM Model by Dataset')
plt.xlabel('LLM Model')
plt.ylabel('RMSE')
plt.xticks(rotation=45)
plt.legend(title='Dataset')
plt.tight_layout()
plt.show()
```
