```python
# Example DataFrame
import pandas as pd
data = {
    'A': [1, 2, 3],
    'B': [1, 2, 3],  # This column is a duplicate of 'A'
    'C': [4, 5, 6],
    'D': [4, 5, 6]   # This column is a duplicate of 'C'
}

df = pd.DataFrame(data)

# Transpose the DataFrame, drop duplicates, then transpose it back to the original shape
df_cleaned = df.T.drop_duplicates().T

print(df_cleaned)
```
